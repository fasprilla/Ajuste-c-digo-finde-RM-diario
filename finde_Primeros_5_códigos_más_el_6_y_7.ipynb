```python
#Inicio
import re
def clean_id(id):
    #Convierte todo a string incluso si el tipo de dato es numerico
    id_str = str(id) if isinstance(id, (int, float)) else id

    # Nos aseguramos que phone_str es tipo string
    if not isinstance(id_str, str):
        return 0

    # Quitamos cualquier caracter no numerico
    id_clean = re.sub('[^0-9]', '', id_str)

    # Verificamos si el ID ahora efectivamente tiene digitos
    if id_clean.isdigit():  # Check if the cleaned ID consists only of digits
        return id_clean  # Convert the cleaned ID string to an integer
    else:
        return 0  # Return None if the cleaned ID is not entirely numeric

def clean_phone_number(phone):

    #Convierte todo a string incluso si el tipo de dato es numerico
    phone_str = str(phone) if isinstance(phone, (int, float)) else phone

    # Nos aseguramos que phone_str es tipo string
    if not isinstance(phone_str, str):
        return 0

    # Quitamos cualquier caracter no numerico
    phone_clean = re.sub('[^0-9]', '', phone_str)

    # Removemos el codigo del pais Colombia si es que se encuentra si la longitud del celular es mayor a 10
    if phone_clean.startswith('57') and len(phone_clean) > 10:
        phone_clean = phone_clean[2:]

    # Verificamos si el celular ahora efectivamente tiene 10 digitos
    if len(phone_clean) == 10:
        return np.int64(phone_clean)  # Convert the cleaned phone number to numpy.int64 format
    else:
        return 0
```

Mover todos los archivos de Bases_hoy a la base de registro del día anterior


```python
import os
import shutil
import datetime

# ------------------------------------------------------------------
# 1. Configuración de rutas base
# ------------------------------------------------------------------

# Carpeta donde llegan las bases de hoy (origen de los archivos a mover/copiar)
ruta_origen = "/content/drive/My Drive/Bases_hoy"

# Diccionario para traducir el número de mes (1-12) a su nombre en español,
# necesario porque las carpetas de archivo están nombradas como "Julio 2026".
MESES_ES = {
    1: "Enero", 2: "Febrero", 3: "Marzo", 4: "Abril",
    5: "Mayo", 6: "Junio", 7: "Julio", 8: "Agosto",
    9: "Septiembre", 10: "Octubre", 11: "Noviembre", 12: "Diciembre"
}


def ruta_carpeta_dia(fecha):
    """
    Dada una fecha (objeto datetime.date), devuelve la ruta de la carpeta
    de archivo que le corresponde, con el formato:
    .../bases_registro_de_bases_antiguas/<Mes Año>/<día>

    Ejemplo: 14 de julio de 2026 -> ".../Julio 2026/14"

    Se usa una función (en vez de una ruta fija) porque el mes puede
    cambiar si el día anterior o el sábado anterior caen en el mes pasado
    (ej. el 1 de un mes, donde "ayer" fue el último día del mes anterior).
    """
    nombre_mes = f"{MESES_ES[fecha.month]} {fecha.year}"
    return os.path.join(
        "/content/drive/My Drive/Re - marketing/bases_registro_de_bases_antiguas",
        nombre_mes,
        str(fecha.day)
    )


# ------------------------------------------------------------------
# 2. Calcular las fechas de hoy y de ayer, y verificar si la carpeta
#    de ayer existe (esto indica si el script corrió normalmente ayer)
# ------------------------------------------------------------------

hoy = datetime.date.today()
ayer = hoy - datetime.timedelta(days=1)
ruta_ayer = ruta_carpeta_dia(ayer)

# ------------------------------------------------------------------
# 3. Decidir el flujo según exista o no la carpeta del día anterior
# ------------------------------------------------------------------

if os.path.isdir(ruta_ayer):
    # --------------------------------------------------------------
    # CASO NORMAL: la carpeta de ayer existe, es decir, no hubo
    # festivo/interrupción. Se MUEVEN los archivos de Bases_hoy a la
    # carpeta del día ANTERIOR (ruta_ayer), no a la de hoy.
    # --------------------------------------------------------------
    ruta_destino = ruta_ayer

    # Recorrer todo lo que hay dentro de la carpeta origen
    for nombre in os.listdir(ruta_origen):
        ruta_item = os.path.join(ruta_origen, nombre)

        # Solo mover archivos, ignorar subcarpetas
        if os.path.isfile(ruta_item):
            shutil.move(ruta_item, os.path.join(ruta_destino, nombre))
            print(f"Movido: {nombre}")

    print(f"✅ Archivos movidos a: {ruta_destino}")

else:
    # --------------------------------------------------------------
    # CASO FESTIVO: no existe la carpeta de ayer (ej. hoy es martes
    # y ayer, lunes, fue festivo, por lo que el script no corrió y
    # nunca se creó esa carpeta).
    #
    # En este caso, en vez de usar la carpeta de hoy, se busca la
    # carpeta del sábado más reciente y se MUEVEN (no copian) los
    # archivos ahí.
    # --------------------------------------------------------------

    # Partir de "ayer" y retroceder día por día hasta encontrar un sábado.
    # weekday() devuelve: lunes=0, martes=1, ... sábado=5, domingo=6
    fecha_sabado = ayer
    while fecha_sabado.weekday() != 5:  # 5 = sábado
        fecha_sabado -= datetime.timedelta(days=1)

    ruta_destino = ruta_carpeta_dia(fecha_sabado)

    # Por si acaso esa carpeta tampoco existiera todavía, se crea
    os.makedirs(ruta_destino, exist_ok=True)

    print(f"⚠️ No existe la carpeta del día anterior ({ruta_ayer}).")
    print(f"Se usará la carpeta del sábado anterior: {ruta_destino}")

    # Recorrer todo lo que hay dentro de la carpeta origen
    for nombre in os.listdir(ruta_origen):
        ruta_item = os.path.join(ruta_origen, nombre)

        # Solo mover archivos, ignorar subcarpetas
        if os.path.isfile(ruta_item):
            shutil.move(ruta_item, os.path.join(ruta_destino, nombre))
            print(f"Movido: {nombre}")

    print(f"✅ Archivos movidos a: {ruta_destino}")


#Para ver que archivos y bases de datos están almacenadas en Bases_Hoy
%cd /content/drive/My Drive/Bases_hoy
%ls
```

### Funcion de Filtrado de Bases

```python
import re

def clean_phone_number(phone):

    #Convierte todo a string incluso si el tipo de dato es numerico
    phone_str = str(phone) if isinstance(phone, (int, float)) else phone

    # Nos aseguramos que phone_str es tipo string
    if not isinstance(phone_str, str):
        return None

    # Quitamos cualquier caracter no numerico
    phone_clean = re.sub('[^0-9]', '', phone_str)

    # Removemos el codigo del pais Colombia si es que se encuentra si la longitud del celular es mayor a 10
    if phone_clean.startswith('57') and len(phone_clean) > 10:
        phone_clean = phone_clean[2:]

    # Verificamos si el celular ahora efectivamente tiene 10 digitos
    if len(phone_clean) == 10:
        return np.int64(phone_clean)  # Convert the cleaned phone number to numpy.int64 format
    else:
        return None

def clean_id(id):
    #Convierte todo a string incluso si el tipo de dato es numerico
    id_str = str(id) if isinstance(id, (int, float)) else id

    # Nos aseguramos que phone_str es tipo string
    if not isinstance(id_str, str):
        return None

    # Quitamos cualquier caracter no numerico
    id_clean = re.sub('[^0-9]', '', id_str)

    # Verificamos si el ID ahora efectivamente tiene digitos
    if id_clean.isdigit():  # Check if the cleaned ID consists only of digits
        return int(id_clean)  # Convert the cleaned ID string to an integer
    else:
        return None  # Return None if the cleaned ID is not entirely numeric
def limpiar_nombre(nombre):

    # Si es null → vacío
    if pd.isna(nombre):
        return ""

    nombre = str(nombre).strip()

    # Eliminar emojis y caracteres raros
    nombre = re.sub(r'[^\w\sÁÉÍÓÚáéíóúÑñ]', '', nombre)

    # Eliminar números
    nombre = re.sub(r'\d+', '', nombre)

    # Quitar espacios dobles
    nombre = re.sub(r'\s+', ' ', nombre).strip()

    # Si queda vacío o muy corto
    if len(nombre) < 3:
        return "Camellador"

    # Tomar solo el primer nombre
    nombre = nombre.split()[0]

    # Capitalize
    return nombre.capitalize()


# =========================
# 🔹 LIMPIAR CELULAR
# =========================
def clean_phone_number(phone):

    if pd.isna(phone):
        return None

    # Dejar solo números
    phone = re.sub(r"\D", "", str(phone))

    # Quitar prefijo 57
    if phone.startswith("57"):
        phone = phone[2:]

    # Validar celular colombiano
    if len(phone) == 10 and phone.startswith("3"):
        return int(phone)

    return None
```

```python
import re

def clean_phone_number(phone):

    #Convierte todo a string incluso si el tipo de dato es numerico
    phone_str = str(phone) if isinstance(phone, (int, float)) else phone

    # Nos aseguramos que phone_str es tipo string
    if not isinstance(phone_str, str):
        return None

    # Quitamos cualquier caracter no numerico
    phone_clean = re.sub('[^0-9]', '', phone_str)

    # Removemos el codigo del pais Colombia si es que se encuentra si la longitud del celular es mayor a 10
    if phone_clean.startswith('57') and len(phone_clean) > 10:
        phone_clean = phone_clean[2:]

    # Verificamos si el celular ahora efectivamente tiene 10 digitos
    if len(phone_clean) == 10:
        return int(phone_clean)  # Convert the cleaned phone number to numpy.int64 format
    else:
        return None

def clean_id(id):
    #Convierte todo a string incluso si el tipo de dato es numerico
    id_str = str(id) if isinstance(id, (int, float)) else id

    # Nos aseguramos que phone_str es tipo string
    if not isinstance(id_str, str):
        return None

    # Quitamos cualquier caracter no numerico
    id_clean = re.sub('[^0-9]', '', id_str)

    # Verificamos si el ID ahora efectivamente tiene digitos
    if id_clean.isdigit():  # Check if the cleaned ID consists only of digits
        return int(id_clean)  # Convert the cleaned ID string to an integer
    else:
        return None  # Return None if the cleaned ID is not entirely numeric
```

```python
# Schedule the campaigns
today_date = datetime.now().strftime('%Y-%m-%d')
scheduled_dates = schedules_dates(today_date)
scheduled_max = f"{today_date}T15:00:00-18:00"   # Cuando envia por la tarde este shedule max (el -18:00 no resta la hora que se seleccione)

# Iterate through the 9 campaigns
for i in range(9):
    date_campaign = i % 3  # ayer -> 0, hace2dias -> 1, hace3dias -> 2
    name_campaign = i // 3  # claro -> 0, nequi -> 1, general -> 2
    # Load the CSV file
    csv = campaigns["csv"][date_campaign + name_campaign * 3]
    df = pd.read_csv(csv)

    # Create the payload to create the campaign
    payload_campaign["campaignNameOriginal"] = campaigns["campaignNameOriginal"][name_campaign]
    payload_campaign["campaignProductConfiguration"]["shortUrl"]["specificUrl"] = campaigns["specificUrl"][name_campaign]
    payload_campaign["campaignProductConfiguration"]["message"] = campaigns["message"][date_campaign]
    payload_campaign["campaignMaxDate"] = scheduled_max
    payload_campaign["scheduledDate"] = scheduled_dates[i]

    response = create_campaign(payload_campaign, headers)
    # Extract the campaignId from the JSON data
    payload_users = create_users_batch(df)
    campaignId = response.json().get('campaignId')
    upload_users(payload_users, headers, campaignId)
    send_message(campaignId, headers)
    print(payload_campaign)
```
